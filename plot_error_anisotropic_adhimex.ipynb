```python
import re
import numpy as np
import matplotlib.pyplot as plt

# ---- Collect the testcases ----
# A testcase is every variable named l2_<case>_<res> WITHOUT 'anis' in it,
# e.g. l2_constantuv_sine_unif_ny640dt0_4nt1 -> case "constantuv_sine_unif", res "ny640dt0_4nt1".
# For each testcase the three bars are:
#   neither: l2_<case>_<res>
#   old:     old_l2_<case>_anis_<res>
#   anis:    l2_<case>_anis_<res>
testcases = []
for name in list(globals()):
    m = re.fullmatch(r"l2_(.+_unif)_(ny\d+dt.+nt\d+)", name)
    if m and "anis" not in name:
        testcases.append((m.group(1), m.group(2)))

neither = [globals()[f"l2_{c}_{r}"] for c, r in testcases]
old     = [globals()[f"old_l2_{c}_anis_{r}"] for c, r in testcases]
anis    = [globals()[f"l2_{c}_anis_{r}"] for c, r in testcases]

# Label on the left of each group, e.g. "constantuv, ny640dt0_4nt1"
labels = [f"{c.replace('_sine_unif', '')}, {r}" for c, r in testcases]

# ---- Plot ----
n = len(testcases)
bar_h = 0.26
y = np.arange(n)

fig, ax = plt.subplots(figsize=(8, 0.55 * n + 1.5))
ax.barh(y - bar_h, neither, height=bar_h, label="iso", color="C0")
ax.barh(y,         old,     height=bar_h, label="old aniso",     color="C1")
ax.barh(y + bar_h, anis,    height=bar_h, label="new aniso",    color="C2")

ax.set_xscale("log")
ax.set_yticks(y)
ax.set_yticklabels(labels)
ax.invert_yaxis()  # first testcase at the top
ax.set_xlabel("$l_2$ error")
ax.grid(True, axis="x", which="both", ls="--", alpha=0.5)
ax.legend()

plt.tight_layout()
plt.show()
```
